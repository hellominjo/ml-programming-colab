(1: 빨간 사각형, 2: 초록 오각형, 3: 파란 육각형)
- 1을 입력 받으면 빨간 사각형
- 2을 입력 받으면 초록 오각형
- 3을 입력 받으면 파란 육각형

### STEP 1 — 준비

```python
# ============================================================
#  실습 도우미  (이 셀을 가장 먼저 한 번만 실행하세요)
# ============================================================
import math, random, sys, time
import matplotlib.pyplot as plt

print("파이썬 버전 :", sys.version.split()[0])
print("준비 완료!")

# ── 미니 거북이 (turtle 설치가 안 될 때 쓰는 예비 도구) ──────────
class MiniTurtle:
    """ColabTurtle 이 설치되지 않아도 도형을 그릴 수 있는 간단한 거북이."""
    def __init__(self):
        self.x = 0.0; self.y = 0.0
        self.angle = 0.0          # 0도 = 오른쪽, 반시계 방향이 +
        self.pen = True
        self.col = "black"
        self.segs = []
    def forward(self, d):
        nx = self.x + d * math.cos(math.radians(self.angle))
        ny = self.y + d * math.sin(math.radians(self.angle))
        if self.pen:
            self.segs.append(((self.x, self.y), (nx, ny), self.col))
        self.x, self.y = nx, ny
    def backward(self, d):  self.forward(-d)
    def right(self, a):     self.angle -= a
    def left(self, a):      self.angle += a
    def color(self, c):     self.col = c
    def penup(self):        self.pen = False
    def pendown(self):      self.pen = True
    def show(self, title="MiniTurtle"):
        fig, ax = plt.subplots(figsize=(4.2, 4.2))
        for (p, q, c) in self.segs:
            ax.plot([p[0], q[0]], [p[1], q[1]], color=c, linewidth=2.4)
        ax.set_aspect("equal"); ax.axis("off"); ax.set_title(title)
        plt.show()

print("MiniTurtle 준비 완료")
```

```python
# ColabTurtle 설치를 시도하고, 안 되면 MiniTurtle 로 자동 전환합니다.
USE_COLAB_TURTLE = False
try:
    import ColabTurtle.Turtle as ct
    USE_COLAB_TURTLE = True
except ImportError:
    try:
        import subprocess, sys
        subprocess.run([sys.executable, "-m", "pip", "-q", "install", "ColabTurtle"],
                       check=True, timeout=90)
        import ColabTurtle.Turtle as ct
        USE_COLAB_TURTLE = True
    except Exception:
        USE_COLAB_TURTLE = False

print("ColabTurtle 사용" if USE_COLAB_TURTLE else "MiniTurtle 사용 (설치 없이 동작)")

def 거북이():
    """강의자료와 같은 방식으로 거북이를 하나 준비합니다."""
    if USE_COLAB_TURTLE:
        try:
            ct.initializeTurtle(initial_speed=5)
        except TypeError:          # 최신 버전 호환
            ct.initializeTurtle()
            ct.speed(5)
        return ct
    return MiniTurtle()
```

###STEP 2 — 도형 그리기

```python
# 사용자에게 번호 입력 받기
choice = int(input("도형을 선택하세요 (1: 빨간 사각형, 2: 초록 오각형, 3: 파란 육각형) : "))

if choice == 1:
    n = 4
    color = "red"

elif choice == 2:
    n = 5
    color = "green"

elif choice == 3:
    n = 6
    color = "blue"

else:
    print("1, 2, 3 중에서 입력해주세요.")
    n = 0


# 도형 그리기
if n > 0:
    t = 거북이()
    t.color(color)

    for _ in range(n):
        t.forward(80)
        t.right(360 / n)

    if not USE_COLAB_TURTLE:
        t.show(f"Regular {n}-gon")
```
